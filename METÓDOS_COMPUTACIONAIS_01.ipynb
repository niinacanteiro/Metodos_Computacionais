ATIVIDADE DE METÓDOS COMPUTACIONAIS: REALIZAÇÃO DE ERRO PERCENTUAL

```python
##APLICAMOS EM TRÊS PONTOS PARA DETERMINAR O VALOR DO PONTO VERDADEIRO.

x = 2
x0 = 1
x1 = 4
x2 = 6

##FUNÇÕES DE X

y0 = 0
y1 = 1.386294
y2 = 1.791759
```

```python
##APLICANDO INTERPOLAÇÃO QUADRADA

b0 = y0
b1 = (y1 - y0) / (x1 - x0)
b2 = (((y2 - y1) / (x2 - x1)) - ((y1 - y0) / (x1 - x0))) / (x2 - x0)

print("valor de b0 é", b0)
print("valor de b1 é", b1)
print("valor de b2 é", b2)
```

```python
##FUNÇÃO APLICADA NA FORMÚLA

def f2(x):
    return b0 + b1 * (x - x0) + b2 * (x - x0) * (x - x1)

print("f2(1) =", f2(x0))
print("f2(4) =", f2(x1))
print("f2(6) =", f2(x2))
print("f2(2) =", f2(x))
```

```python
##DECLARAÇÃO DE VARIÁVEIS

import math
real = math.log(x)
aproximado = f2(x)

print(real)
print(aproximado)

E = ((real-aproximado)/real)
print(E)
```

```python
##ERRO PERCENTUAL

print("Erro percentual",E*100,"%")
```
